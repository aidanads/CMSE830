```python
eda2 = clean_df[["Height", "Volume_Rep", "Main_P_Map"]].dropna()

print("Observations:", len(eda2))

# Check distributions and extreme values
display(
    eda2[["Height", "Volume_Rep"]]
    .describe(percentiles=[.5, .9, .99])
    .T
)
# Largest reservoir capacities
display(eda2.nlargest(5, "Volume_Rep"))
```

```python
# Keep positive volumes for logarithmic visualization
plot_df = eda2[eda2["Volume_Rep"] > 0].copy()

plt.figure(figsize=(11, 6))

sns.scatterplot(
    data=plot_df,
    x="Height",
    y="Volume_Rep",
    hue="Main_P_Map",
    alpha=0.4,
    s=25,
    palette="tab10",
    edgecolor="black",  
    linewidth=0.3
)

plt.yscale("log")
plt.title("Dam Height vs. Reservoir Capacity by Purpose")
plt.xlabel("Dam Height (m)")
plt.ylabel("Reservoir Capacity (million m³, log scale)")
plt.tight_layout()
plt.show()

# Statistical summary
print("Observations:", len(plot_df))
print("Spearman correlation:",
      round(plot_df["Height"].corr(plot_df["Volume_Rep"], method="spearman"), 3))
```

```python
# Correlation by dam purpose
purpose_corr = plot_df.groupby("Main_P_Map")[["Height", "Volume_Rep"]].apply(
    lambda x: x["Height"].corr(x["Volume_Rep"], method="spearman")
)

purpose_count = plot_df.groupby("Main_P_Map").size()

summary2 = pd.DataFrame({
    "Count": purpose_count,
    "Spearman correlation": purpose_corr
}).sort_values("Count", ascending=False)

display(summary2.round(3))
```

**Question 2 interpretation and limitation: I created a scatterplot using 14,009 observations to examine the relationship between dam height and reservoir capacity. I used a logarithmic scale for capacity because the distribution is strongly right-skewed, with a median of 6.9 and a mean of 1,209.4 million m³. I calculated Spearman correlation because it is less sensitive to extreme values. The overall correlation (0.543) indicates a moderate positive relationship, meaning taller dams generally have larger reservoir capacities. However, the relationship varies by purpose: Hydroelectricity (0.539), Irrigation (0.542), and Water Supply (0.522) show similar moderate correlations, while Recreation (-0.021) shows almost no relationship and Livestock (0.841) shows a strong positive relationship.**

*Reservoir capacity is missing for 55.48% of the dataset, and missingness varies across dam purposes. This may introduce selection bias, so the observed relationships do not represent all dams equally.*
